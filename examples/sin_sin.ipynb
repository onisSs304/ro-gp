```python
import autograd.numpy as anp
import pymanopt
from pymanopt.manifolds import Sphere, Product
import numpy as np
from pymanopt.tools import diagnostics
import matplotlib.pyplot as plt
from scipy.optimize import minimize
from autograd import value_and_grad
```

```python
manifold = Sphere(3)
@pymanopt.function.autograd(manifold)
def cost_mul(point):
    return point[1]

problem_mul = pymanopt.Problem(manifold, cost_mul)
```

```python
for i in range(10):
    X0 = problem_mul.manifold.random_point()
    optimizer_mul = pymanopt.optimizers.SteepestDescent(verbosity=0)
    result_mul = optimizer_mul.run(problem_mul, initial_point=X0)
    print(f"Run for mul. sin-sin {i+1}: {result_mul.point}, {result_mul.cost} ")
```
